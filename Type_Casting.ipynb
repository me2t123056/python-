**Type Casting** means converting one data type into another.

int → float,
float → int,
int → char,
char → int.

```python
#int → float
a=12
result=float(a)
print(result)
```

```python
#float → int
a=12.89
result1=int(a)
print(result1) # data loss

b=12.00
result2=int(b)
print(result2) #No data loss
```

```python
#int → char
#ASCII code to a character
a=65
result=chr(a)
print(result)
```

```python
#char → int
#character into its Unicode code
name="M" # only single character possible for char -> int conversion
result=ord(name) # Use ord() to get the Unicode integer value
print(result)
```
