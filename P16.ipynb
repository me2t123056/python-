```python
import mysql.connector
try:
    # Connect to the database inside Colab
con = mysql.connector.connect(
        host="localhost",
        user="root",
        password="your_password",
database="Clinic_DB"
    )
    cursor = con.cursor()
    # INSERT
    insert_query = (
        "INSERT INTO Patients (Name, Ailment) "
        "VALUES (%s, %s)"
    )
cursor.execute(insert_query, ("Meet Italiya", "Fever"))
    con.commit()

    print(f"Inserted {cursor.rowcount} record(s).")
    update_query = (
        "UPDATE Patients SET Ailment = %s "
        "WHERE Name = %s"
    )
cursor.execute(update_query, ("Recovered", "Meet Italiya"))
    con.commit()
    print(f"Updated {cursor.rowcount} record(s).")
    cursor.execute("SELECT * FROM Patients")
    results = cursor.fetchall()
    for row in results:
        print("Record:", row)
    delete_query = "DELETE FROM Patients WHERE Name = %s"
cursor.execute(delete_query, ("Meet Italiya",))
    con.commit()
    print(f"Deleted {cursor.rowcount} record(s).")
except mysql.connector.Error as err:
    print(f"Database Error: {err}")
finally:
    if "con" in locals() and con.is_connected():
        cursor.close()
        con.close()
        print("Database connection closed.")

```
